```python
import csv


class Book:
    def __init__(self, title, author, genre, publication_year):
        self.title = title
        self.author = author
        self.genre = genre
        self.publication_year = publication_year
class Library:
    def __init__(self):
        # List to store Book objects
        self.lib = []
        self.load_from_file()
    def load_from_file(self, filename="library.csv"):
      try:
        with open(filename, mode="r", newline="", encoding="utf-8") as file:
            reader = csv.DictReader(file)

            for row in reader:
                book = Book(
                    row["title"],
                    row["author"],
                    row["genre"],
                    int(row["publication_year"])
                )
                self.lib.append(book)

      except FileNotFoundError:
        # Expected case → first run
        print("No saved library found. Starting with an empty library.")

      except ValueError:
        # Data corrupted
        print("Error: Invalid data format in file.")

      except Exception as e:
        # Unexpected error
        print(f"Unexpected error while loading data: {e}")


    def save_to_file(self, filename="library.csv"):
      try:
        with open(filename, mode="w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["title", "author", "genre", "publication_year"])

            for book in self.lib:
                writer.writerow([
                    book.title,
                    book.author,
                    book.genre,
                    book.publication_year
                ])

      except PermissionError:
        print("Error: File permission denied. Close the file and try again.")

      except Exception as e:
        print(f"Failed to save data: {e}")


    def add_book(self, book):
        # Add a Book object to the library
        self.lib.append(book)
        self.save_to_file()


    def remove_book(self, title, author):
        for book in self.lib:
            if book.title == title and book.author == author:
                self.lib.remove(book)
                self.save_to_file()
                print("Book removed successfully")
                return   # stop execution once removed

        print("Book not found")


    def search_book(self, title, author):
        for book in self.lib:
            if book.title == title and book.author == author:
                print("\nBook found:")
                print(f"Title: {book.title}")
                print(f"Author: {book.author}")
                print(f"Genre: {book.genre}")
                print(f"Year: {book.publication_year}")
                return

        print("Book not found")


    def display_library(self):
        # Check if library is empty
        if not self.lib:
            print("Library is empty")
            return

        # Display all books
        for i, book in enumerate(self.lib, start=1):
            print(f"\nBook {i}")
            print(f"Title: {book.title}")
            print(f"Author: {book.author}")
            print(f"Genre: {book.genre}")
            print(f"Year: {book.publication_year}")

def get_int_input(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Please enter a valid number.")

def main():
    my_library = Library()

    while True:
        print("\n" + "="*40)
        print("    Personal Library Management System")
        print("="*40)
        try:
            choice = get_int_input(
                "\n1- Add\n2- Remove\n3- Search\n4- Display\n5- Exit\n"
            )
            if choice == 1:
              title = input("Title: ")
              author = input("Author: ")
              genre = input("Genre: ")
              year = int(input("Publication Year: "))

              book = Book(title, author, genre, year)
              my_library.add_book(book)

            elif choice == 2:
              title = input("Title: ")
              author = input("Author: ")
              my_library.remove_book(title, author)

            elif choice == 3:
              title = input("Title: ")
              author = input("Author: ")
              my_library.search_book(title, author)

            elif choice == 4:
              my_library.display_library()

            elif choice == 5:
              break

        except KeyboardInterrupt:
            print("\nProgram interrupted by user.")
            break
main()
```

```python
import csv
import os
from typing import Optional, List


class Book:
    def __init__(self, title: str, author: str, genre: str, publication_year: int):
        self.title = title.strip()
        self.author = author.strip()
        self.genre = genre.strip()
        self.publication_year = publication_year

    def __str__(self) -> str:
        return f"{self.title} | {self.author} | {self.genre} | {self.publication_year}"

    def to_dict(self) -> dict:
        return {
            "title": self.title,
            "author": self.author,
            "genre": self.genre,
            "publication_year": self.publication_year
        }

    @classmethod
    def from_dict(cls, data: dict) -> "Book":
        return cls(
            title=data["title"],
            author=data["author"],
            genre=data["genre"],
            publication_year=int(data["publication_year"])
        )


class Library:
    def __init__(self, filename: str = "library.csv"):
        self.filename = filename
        self.books: List[Book] = []
        self.load_from_file()

    def load_from_file(self) -> None:
        """تحميل البيانات من الملف بأمان مع معالجة الأخطاء"""
        if not os.path.exists(self.filename):
            print(f"ملف {self.filename} غير موجود، سيتم إنشاؤه عند الحفظ.")
            return

        try:
            with open(self.filename, mode="r", encoding="utf-8", newline="") as file:
                reader = csv.DictReader(file)
                for row in reader:
                    try:
                        book = Book.from_dict(row)
                        self.books.append(book)
                    except (KeyError, ValueError) as e:
                        print(f"سطر تالف في الملف: {row} → {e}")

        except PermissionError:
            print(f"خطأ: لا يوجد صلاحية لقراءة الملف {self.filename}")
        except UnicodeDecodeError:
            print(f"خطأ: ترميز الملف {self.filename} غير مدعوم (جرب utf-8)")
        except Exception as e:
            print(f"خطأ غير متوقع أثناء قراءة الملف: {e}")

    def save_to_file(self) -> None:
        """حفظ البيانات بأمان مع معالجة الأخطاء"""
        try:
            with open(self.filename, mode="w", encoding="utf-8", newline="") as file:
                writer = csv.DictWriter(file, fieldnames=["title", "author", "genre", "publication_year"])
                writer.writeheader()
                for book in self.books:
                    writer.writerow(book.to_dict())
        except PermissionError:
            print(f"خطأ: لا يمكن الكتابة في الملف {self.filename} (تحقق من الصلاحيات)")
        except Exception as e:
            print(f"فشل حفظ المكتبة: {e}")

    def add_book(self, book: Book) -> None:
        """إضافة كتاب مع التحقق من التكرار"""
        if any(b.title == book.title and b.author == book.author for b in self.books):
            print(f"الكتاب '{book.title}' لـ '{book.author}' موجود بالفعل!")
            return

        self.books.append(book)
        self.save_to_file()
        print(f"تم إضافة الكتاب بنجاح: {book}")

    def remove_book(self, title: str, author: str) -> bool:
        title = title.strip()
        author = author.strip()

        for i, book in enumerate(self.books):
            if book.title == title and book.author == author:
                removed = self.books.pop(i)
                self.save_to_file()
                print(f"تم حذف الكتاب: {removed}")
                return True

        print("الكتاب غير موجود.")
        return False

    def search_book(self, title: str, author: str) -> None:
        title = title.strip()
        author = author.strip()

        found = False
        for book in self.books:
            if book.title == title and book.author == author:
                print("\nتم العثور على الكتاب:")
                print(f"العنوان   : {book.title}")
                print(f"المؤلف    : {book.author}")
                print(f"التصنيف   : {book.genre}")
                print(f"سنة النشر : {book.publication_year}")
                found = True
                break  # نفترض أن العنوان + المؤلف مفتاح فريد

        if not found:
            print("الكتاب غير موجود.")

    def display_library(self) -> None:
        if not self.books:
            print("المكتبة فارغة حالياً.")
            return

        print(f"\nالمكتبة تحتوي على {len(self.books)} كتاب/كتب:")
        print("-" * 60)
        for i, book in enumerate(self.books, 1):
            print(f"{i:2d}. {book}")
        print("-" * 60)


def get_valid_year(prompt: str) -> int:
    """دالة مساعدة لأخذ سنة نشر صحيحة"""
    while True:
        try:
            year = int(input(prompt))
            if 0 < year <= 2100:  # نطاق منطقي
                return year
            print("السنة يجب أن تكون بين 1 و 2100.")
        except ValueError:
            print("يرجى إدخال رقم صحيح.")


def main():
    my_library = Library()

    while True:
        print("\n" + "="*40)
        print("     نظام إدارة المكتبة الشخصية")
        print("="*40)
        print("1. إضافة كتاب")
        print("2. حذف كتاب")
        print("3. البحث عن كتاب")
        print("4. عرض جميع الكتب")
        print("5. خروج")
        print("="*40)

        try:
            choice = int(input("اختر (1-5): ").strip())

            if choice == 1:
                title = input("العنوان: ").strip()
                author = input("المؤلف: ").strip()
                genre = input("التصنيف: ").strip()
                year = get_valid_year("سنة النشر: ")

                if not title or not author:
                    print("العنوان والمؤلف مطلوبان!")
                    continue

                book = Book(title, author, genre, year)
                my_library.add_book(book)

            elif choice == 2:
                title = input("العنوان: ").strip()
                author = input("المؤلف: ").strip()
                my_library.remove_book(title, author)

            elif choice == 3:
                title = input("العنوان: ").strip()
                author = input("المؤلف: ").strip()
                my_library.search_book(title, author)

            elif choice == 4:
                my_library.display_library()

            elif choice == 5:
                print("شكراً لاستخدام النظام. مع السلامة!")
                break

            else:
                print("اختيار غير صحيح، حاول مرة أخرى.")

        except ValueError:
            print("يرجى إدخال رقم من 1 إلى 5.")
        except KeyboardInterrupt:
            print("\nتم إيقاف البرنامج بواسطة المستخدم.")
            break
        except Exception as e:
            print(f"حدث خطأ غير متوقع: {e}")


if __name__ == "__main__":
    main()
```
